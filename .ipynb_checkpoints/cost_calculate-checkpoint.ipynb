```python
import re
import math
import os
import pandas as pd

# ==========================================
# 1. Data Parser (.dat file)
# ==========================================
def parse_dat_file(dat_filepath):
    params = {}
    try:
        if not os.path.exists(dat_filepath):
            print(f"エラー: ファイルが見つかりません -> {dat_filepath}")
            return None

        with open(dat_filepath, 'r', encoding='utf-8') as f:
            content = f.read()

        # --- Helper functions for regex ---
        def get_scalar(name, type_func=float):
            # Matches "name = value;" pattern
            pattern = re.compile(rf'{name}\s*=\s*([-+]?(?:[0-9]*\.?[0-9]+)(?:[eE][-+]?[0-9]+)?)\s*;')
            match = pattern.search(content)
            if match:
                return type_func(match.group(1))
            return 0.0

        def get_array(name, type_func=float):
            # Matches "name = [v1, v2, ...];" pattern
            pattern = re.compile(rf'{name}\s*=\s*\[(.*?)\];', re.DOTALL)
            match = pattern.search(content)
            if match:
                raw_vals = match.group(1).replace('\n', ' ').split(',')
                return [type_func(v.strip()) for v in raw_vals if v.strip()]
            return []

        # --- Basic Parameters ---
        params['scaleLength'] = get_scalar('scaleLength')
        params['stringSpacing'] = get_scalar('stringSpacing')
        params['timePerMm'] = get_scalar('timePerMm')
        
        # Arrays
        params['fingerLengths_mm'] = get_array('fingerLengths_mm') 
        params['fingerDifficulty'] = get_array('fingerDifficulty') 
        params['natural_min_spread_mm'] = get_array('natural_min_spread_mm')
        params['SP_max_mm'] = get_array('SP_max_mm')
        params['availableTime'] = get_array('availableTime', float) # OPL uses float for time
        
        # Scalars
        params['timeThresholdRatio'] = get_scalar('timeThresholdRatio')
        params['highPositionPenalty'] = get_scalar('highPositionPenalty')
        params['s_max'] = int(get_scalar('s_max'))
        params['f_max'] = int(get_scalar('f_max'))
        params['p_max'] = int(get_scalar('p_max'))

        # Cost Weights
        params['d_PC'] = get_scalar('d_PC')
        params['d_SC'] = get_scalar('d_SC')
        params['d_HS'] = get_scalar('d_HS')
        params['d_DH'] = get_scalar('d_DH')
        params['d_NO'] = get_scalar('d_NO')

        params['alpha_PC'] = get_scalar('alpha_PC')
        params['alpha_SC'] = get_scalar('alpha_SC')
        params['alpha_HS'] = get_scalar('alpha_HS')
        params['alpha_DH'] = get_scalar('alpha_DH')
        params['alpha_NO'] = get_scalar('alpha_NO')

        print(f"INFO: パラメータ読み込み完了: {dat_filepath}")
        return params

    except Exception as e:
        print(f"ERROR: .datファイルの読み込みに失敗しました: {e}")
        return None

# ==========================================
# 2. Tablature Parser (.txt file)
# ==========================================
def parse_solution_file(solution_filepath):
    notes_data = []
    try:
        if not os.path.exists(solution_filepath):
            print(f"エラー: ファイルが見つかりません -> {solution_filepath}")
            return []

        with open(solution_filepath, 'r', encoding='utf-8') as f:
            lines = f.readlines()

        header_found = False
        for line in lines:
            line = line.strip()
            if not line: continue
            
            if re.match(r'Note\s+String\s+Fret\s+Handposition\s+Finger', line):
                header_found = True
                continue
            
            if not header_found: continue
            if 'Cost Breakdown' in line or 'Total Cost' in line: break
            if line.startswith('---'): break

            parts = re.split(r'\s+', line)
            if len(parts) >= 5:
                note_name = parts[0]
                try:
                    if note_name == 'R':
                        notes_data.append({
                            'note': 'R', 'string': 0, 'fret': 0, 'hand': 0, 'finger': 0 
                        })
                    else:
                        finger_val = parts[4]
                        finger_num = 0 if finger_val == 'x' else int(finger_val)
                        notes_data.append({
                            'note': note_name,
                            'string': int(parts[1]),
                            'fret': int(parts[2]),
                            'hand': int(parts[3]),
                            'finger': finger_num
                        })
                except ValueError:
                    continue
        
        return notes_data

    except Exception as e:
        print(f"ERROR: ソリューションファイルの読み込みに失敗しました: {e}")
        return []

# ==========================================
# 3. Cost Calculation Logic (Aligned with OPL)
# ==========================================

class CostCalculator:
    def __init__(self, params):
        self.p = params
        
        # --- 1. Pre-calculate Fret Positions (OPL Section 2A) ---
        # Index 0 to f_max
        self.fret_pos = [
            self.p['scaleLength'] - (self.p['scaleLength'] / (2 ** (f / 12.0)))
            for f in range(self.p['f_max'] + 1)
        ]

        # --- 2. Slant Model & Effective Lengths (OPL Section 2B) ---
        # Constants hardcoded in OPL
        self.knuckleOffsets = [0.0, 0.0, 20.0, 40.0, 60.0] # 1-based index padding: [dummy, F1, F2, F3, F4]
        self.maxRotationAngle = 60.0
        
        self.s_max = self.p['s_max'] # usually 6
        
        # Initialize tables
        # NaturalReachTable[string_idx][finger_idx] (using 1-based indexing for convenience in logic)
        self.NaturalReachTable = {}     
        self.EffectiveFingerLength = {} 
        
        # Pad arrays to match 1-based finger indexing of OPL
        # fingerLengths_mm in dat is [F1, F2, F3, F4] (size 4)
        # We map: 1->0, 2->1, 3->2, 4->3
        finger_len = [0.0] + self.p['fingerLengths_mm'] 
        
        # Pad arrays for SP_max and natural_min
        # In OPL: SP_max_mm[FingersButFirst] -> size 3 (fingers 2,3,4)
        # dat file: [60, 80, 100] corresponds to F2, F3, F4
        full_SP_max = [0.0, 0.0] + self.p['SP_max_mm'] # [dummy, dummy(F1=0), F2, F3, F4]
        
        full_natural_min = [0.0, 0.0] + self.p['natural_min_spread_mm'] # [dummy, dummy, F2, F3, F4]

        PI = 3.141592653589793
        
        for s in range(1, self.s_max + 1):
            angleDeg = self.maxRotationAngle * (self.s_max - s) / (self.s_max - 1)
            angleRad = angleDeg * (PI / 180.0)
            
            sinA = math.sin(angleRad)
            cosA = math.cos(angleRad)
            w = angleDeg / self.maxRotationAngle
            
            self.NaturalReachTable[s] = {}
            self.EffectiveFingerLength[s] = {}
            
            for h in range(1, 5): # Fingers 1 to 4
                # Geom Natural
                geomNatural = (self.knuckleOffsets[h] * sinA) + (finger_len[h] * 0.3)
                calcNatural = (1.0 - w) * full_natural_min[h] + w * geomNatural
                
                self.NaturalReachTable[s][h] = calcNatural
                
                # Effective Length (Y-axis)
                self.EffectiveFingerLength[s][h] = finger_len[h] * cosA * 0.9

        # --- 3. HS Cost Table (OPL Section 2C) ---
        self.HS_CostTable = {} # Key: (hand_pos, finger, string, fret) -> cost
        
        # 【修正】 toleranceをインスタンス変数として定義
        self.tolerance = 10.0
        
        # This corresponds to the pre-calculation. 
        # In Python, we can calculate on-the-fly or cache. 
        # To be fast and strict, we'll compute as needed but follow the formula.

    def get_hs_cost(self, hand_pos, finger, note_string, note_fret):
        # Corresponds to HS_CostTable lookup
        if note_fret >= hand_pos:
            actual_dist = self.fret_pos[note_fret] - self.fret_pos[hand_pos]
            ideal_dist = self.NaturalReachTable[note_string][finger]
            raw_diff = abs(actual_dist - ideal_dist)
            
            # 【修正】 self.toleranceを使用
            return max(0.0, raw_diff - self.tolerance)
        else:
            return 100000.0 # Should not happen in valid solution

    def calculate(self, notes_data):
        # Reset totals
        total_PC = 0.0
        total_SC = 0.0
        total_JointSavings = 0.0
        total_HS = 0.0
        total_DH = 0.0
        total_NO = 0.0

        details = []
        num_notes = min(len(notes_data), len(self.p['availableTime']))
        
        # Data preparation for Time Logic (OPL Section 2D)
        # Map notes to 1-based index for easy OPL comparison
        # notes_data[i] corresponds to OPL note[i+1]
        
        prev_note_idx = {} # i -> prev_index
        accumulated_time = {} # i -> time
        
        # Calculate accumulated times and previous indices
        for i in range(num_notes):
            curr_note = notes_data[i]
            is_rest = (curr_note['note'] == 'R')
            
            if not is_rest:
                # Look backwards for the previous non-rest note
                p = i - 1
                time_sum = 0.0
                while p >= 0 and notes_data[p]['note'] == 'R':
                    time_sum += self.p['availableTime'][p]
                    p -= 1
                
                if p >= 0:
                    prev_note_idx[i] = p
                    accumulated_time[i] = time_sum + self.p['availableTime'][p]
                else:
                    # First note or only rests before
                    prev_note_idx[i] = i 
                    accumulated_time[i] = 10000.0
            else:
                accumulated_time[i] = 0.0
                prev_note_idx[i] = -1 # Invalid

        # --- Main Loop ---
        last_played_idx = None

        for i in range(num_notes):
            curr = notes_data[i]
            is_rest = (curr['note'] == 'R')
            
            c_pc = 0.0
            c_sc = 0.0
            c_joint = 0.0
            c_hs = 0.0
            c_dh = 0.0
            c_no = 0.0

            time_penalty = 0.0
            
            if not is_rest:
                # --- Time Penalty Factor ---
                # dexpr float timePenaltyFactor
                time_penalty = min(1.0, 50.0 / (accumulated_time[i] + 0.01))

                # ================================
                # Static Costs (HS, DH, NO)
                # ================================
                
                # NO: Open String
                # Checks if current string is open. 
                # Logic: If fret is 0, it's an open string cost.
                if curr['fret'] == 0:
                    c_no = 1.0

                # HS: Hand Spread
                # OPL: sum(p) sum(h) ... HS_CostTable
                if curr['finger'] in [1, 2, 3, 4] and curr['fret'] > 0: 
                    # Note: In OPL model provided, loop for HS is over 'h in Fingers' (1..4).
                    # However, usually finger 1 defines the hand pos (dist=0), but here we use the table logic.
                    # Table logic: actualDist(fret - hand_pos) vs NaturalReach(string, finger)
                    c_hs = self.get_hs_cost(curr['hand'], curr['finger'], curr['string'], curr['fret'])
                
                # DH: Difficulty Hold
                # OPL: (fingerDifficulty[h] - 1) + maxl(0, reach_shortage)
                if 1 <= curr['finger'] <= 4:
                    f_idx = curr['finger'] - 1
                    
                    term1 = self.p['fingerDifficulty'][f_idx] - 1.0
                    
                    # Reach shortage (Y-axis)
                    # ((c.corda - 1) * stringSpacing) - EffectiveFingerLength[c.corda][h] * 0.8
                    dist_y = (curr['string'] - 1) * self.p['stringSpacing']
                    eff_len = self.EffectiveFingerLength[curr['string']][curr['finger']]
                    term2 = max(0.0, dist_y - (eff_len * 0.8))
                    
                    c_dh = term1 + term2

                # ================================
                # Dynamic Costs (PC, SC)
                # ================================
                prev_idx = prev_note_idx.get(i, i)
                
                if prev_idx != i and prev_idx != -1:
                    prev = notes_data[prev_idx]
                    
                    # PC: Position Change
                    # abs(fretPosition[prev_hand] - fretPosition[curr_hand])
                    dist_pc = abs(self.fret_pos[prev['hand']] - self.fret_pos[curr['hand']])
                    c_pc = dist_pc
                    
                    # SC: String Change
                    # stringSpacing * abs(prev_string - curr_string)
                    # Note: If open string, OPL uses 's' directly.
                    # Here notes_data['string'] is the string played.
                    dist_sc = self.p['stringSpacing'] * abs(prev['string'] - curr['string'])
                    c_sc = dist_sc

                    # Joint Savings
                    # Condition: h==1 (Index finger only), same fret, different string
                    if curr['finger'] == 1 and prev['finger'] == 1:
                        # Check if bar/roll is possible
                        # OPL: c.fret == c_prev.fret && abs(c.corda - c_prev.corda) >= 1
                        if (curr['fret'] == prev['fret']) and (abs(curr['string'] - prev['string']) >= 1):
                             # Discount
                             c_joint = self.p['stringSpacing'] * 0.9

            # ================================
            # Weighting & Totaling
            # ================================
            # Apply Time Penalty to PC, SC, Joint
            w_pc = c_pc * time_penalty * self.p['d_PC'] * self.p['alpha_PC']
            
            # SC Net Cost = (SC - Joint) * factor
            # However, we track them separately for breakdown usually, but OPL formula is:
            # alpha_SC * d_SC * (tot_SC - tot_JointSavings)
            # Let's calculate raw weighted values first
            
            raw_sc_weighted = c_sc * time_penalty * self.p['d_SC'] * self.p['alpha_SC']
            raw_joint_weighted = c_joint * time_penalty * self.p['d_SC'] * self.p['alpha_SC'] # Uses SC weights
            
            w_sc_net = raw_sc_weighted - raw_joint_weighted
            
            w_hs = c_hs * self.p['d_HS'] * self.p['alpha_HS']
            w_dh = c_dh * self.p['d_DH'] * self.p['alpha_DH']
            w_no = c_no * self.p['d_NO'] * self.p['alpha_NO']

            total_PC += w_pc
            total_SC += raw_sc_weighted
            total_JointSavings += raw_joint_weighted
            total_HS += w_hs
            total_DH += w_dh
            total_NO += w_no

            details.append({
                "Index": i+1,
                "Note": curr['note'],
                "PC_cost": w_pc,
                "SC_cost": w_sc_net, # Showing net SC in table
                "HS_cost": w_hs,
                "DH_cost": w_dh,
                "NO_cost": w_no,
                "Total": w_pc + w_sc_net + w_hs + w_dh + w_no
            })

        # Final Totals
        final_SC = total_SC - total_JointSavings
        grand_total = total_PC + final_SC + total_HS + total_DH + total_NO

        return {
            "total": grand_total,
            "breakdown": {
                "PC": total_PC,
                "SC": final_SC, # Net SC
                "HS": total_HS,
                "DH": total_DH,
                "NO": total_NO
            },
            "details": pd.DataFrame(details)
        }

# ==========================================
# 4. Main Execution
# ==========================================

def main():
    datas_dir = "datas"
    results_dir = "results"
    
    # Note: Update filenames as needed based on your upload
    dat_filename = "complex_solo.dat" 
    sol_filename = "complex_solo_modified_model (209).txt" 
    
    # Auto-find logic (optional, keeping from original)
    if not os.path.exists(os.path.join(results_dir, sol_filename)):
         if os.path.exists(results_dir):
            files = [f for f in os.listdir(results_dir) if f.endswith('.txt') and 'modified_model' in f]
            if files:
                sol_filename = files[-1]
    
    # Adjust paths if files are in current directory (for Colab/Notebook)
    if not os.path.exists(os.path.join(datas_dir, dat_filename)):
        dat_path = dat_filename # Try current dir
    else:
        dat_path = os.path.join(datas_dir, dat_filename)

    if not os.path.exists(os.path.join(results_dir, sol_filename)):
        sol_path = sol_filename # Try current dir
    else:
        sol_path = os.path.join(results_dir, sol_filename)
    
    print(f"Reading params from: {dat_path}")
    params = parse_dat_file(dat_path)
    if not params: return

    print(f"Reading solution from: {sol_path}")
    notes_data = parse_solution_file(sol_path)
    if not notes_data: return

    calculator = CostCalculator(params)
    result = calculator.calculate(notes_data)

    print("\n" + "="*50)
    print(" --- Cost Breakdown (OPL Replication) ---")
    bd = result['breakdown']
    print(f"  Position Change (PC): {bd['PC']:.9f}")
    print(f"  String Change   (SC): {bd['SC']:.9f}")
    print(f"  Hand Spread     (HS): {bd['HS']:.9f}")
    print(f"  Difficulty Hold (DH): {bd['DH']:.9f}")
    print(f"  Open String     (NO): {bd['NO']:.9f}")
    print("-" * 40)
    print(f"  Total Cost (Sum)    : {result['total']:.9f}")
    print("="*50)

    print("\n[Details]")
    pd.set_option('display.max_columns', None)
    pd.set_option('display.width', 1000)
    print(result['details'].to_string(index=False))

if __name__ == "__main__":
    main()
```
